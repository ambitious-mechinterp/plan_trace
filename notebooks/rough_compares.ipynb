```python
import json
```

```python
def _load_json(file):
    with open(file, "r") as f:
        data = json.load(f)
    return data
```

```python
cluster_w4 = "../outputs/per_pos/test_w4/prompt_15/token_297/clusters.json"
cluster_w10 = "../outputs/per_pos/test_w10/prompt_15/token_297/clusters.json"

cluster_w4_data = _load_json(cluster_w4)
cluster_w10_data = _load_json(cluster_w10)
```

```python
def compare_dicts(dict1, dict2):
    """
    Compare two dictionaries with nested list structure.
    Returns (are_same, differences)
    """
    differences = {}
    
    # Check if keys match
    keys1 = set(dict1.keys())
    keys2 = set(dict2.keys())
    
    if keys1 != keys2:
        differences['missing_keys'] = {
            'only_in_dict1': keys1 - keys2,
            'only_in_dict2': keys2 - keys1
        }
    
    # Compare common keys
    common_keys = keys1 & keys2
    
    for key in common_keys:
        list1 = dict1[key]
        list2 = dict2[key]
        
        # Convert inner lists to tuples for comparison (making the innermost list a frozenset)
        def normalize_item(item):
            return (item[0], item[1], frozenset(item[2]))
        
        set1 = set(normalize_item(item) for item in list1)
        set2 = set(normalize_item(item) for item in list2)
        
        if set1 != set2:
            differences[key] = {
                'only_in_dict1': [list(item[:2]) + [sorted(item[2])] for item in (set1 - set2)],
                'only_in_dict2': [list(item[:2]) + [sorted(item[2])] for item in (set2 - set1)]
            }
    
    are_same = len(differences) == 0
    
    return are_same, differences
```

```python
are_same, diffs = compare_dicts(cluster_w4_data, cluster_w10_data)

if are_same:
    print("✓ The dictionaries are identical (order-independent)")
else:
    print("✗ The dictionaries are different:")
    print("\nDifferences:")
    for key, diff in diffs.items():
        if key == 'missing_keys':
            if diff['only_in_dict1']:
                print(f"  Keys only in dict1: {diff['only_in_dict1']}")
            if diff['only_in_dict2']:
                print(f"  Keys only in dict2: {diff['only_in_dict2']}")
        else:
            print(f"\n  Key '{key}':")
            if diff['only_in_dict1']:
                print(f"    Only in dict1: {diff['only_in_dict1']}")
            if diff['only_in_dict2']:
                print(f"    Only in dict2: {diff['only_in_dict2']}")
```
