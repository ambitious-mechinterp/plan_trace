```python
import requests

url = "https://www.neuronpedia.org/api/activation/get"
payload = {
    "modelId": "gemma-2-2b",
    "source": "4-gemmascope-transcoder-16k",
    "index": "13154",
}
headers = {"Content-Type": "application/json"}

response = requests.post(url, json=payload, headers=headers, timeout=30)
response.raise_for_status()
print(response.json())
```

```python
len(response.json())
```

```python
response.json()[0]['maxValue']
```

```python
import matplotlib.pyplot as plt

# json_data: your list of 55 dicts you already have
# e.g., json_data = response.json()
max_vals = [item['maxValue'] for item in response.json()]

plt.figure(figsize=(8, 3))
plt.plot(max_vals, marker='o', linewidth=1)
plt.xlabel('index')
plt.ylabel('maxValue')
plt.title('maxValue across items')
plt.tight_layout()
plt.show()
```
